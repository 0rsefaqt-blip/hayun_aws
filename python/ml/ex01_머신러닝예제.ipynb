```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# sklearn 에서 제공하는 붓꽃 데이터
# 붓꽃 정보(꽃받침 길이 (sepal Length), 꽃받침 너비(sepal width), 꽃잎 길이 (petal Length), 꽃잎 너비 (petal width)를 
# 이용하여 붓꽃 종류를 학습시키고 예측
iris = load_iris(as_frame=True)

X = iris.data # feature 특성
y = iris.target # 붓꽃종류

# 훈련 데이터와 테스트 데이터로 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y #, random_state=42
)
```

```python
# 모델 선택 : K 최근접이웃
model = KNeighborsClassifier(n_neighbors=5)
# 학습
model.fit(X_train, y_train)
```

```python

# 테스트 데이터를 가지고 예측
pred = model.predict(X_test)

# 딥과 (y_test)과 예측값을 비교하여 정확도를 확인
print("정확도:", accuracy_score(y_test, pred))   # 1.0
```

```python
# 새로운 데이터
new_df = pd.DataFrame([[5.0, 3.4, 1.5, 0.2], [6.7, 3.0, 5.2, 2.3]], columns=X.columns)

# 새 데이터를 기준으로 예측
p = model.predict(new_df)

# 예측된 값과 품종 이름
print(p, iris.target_names[p])           # [0 2] ['setosa' 'virginica']
print(model.predict_proba(new_df))       # 품종별 확률
```

```python
# 스케일링 예제
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X, y = load_breast_cancer(return_X_y=True, as_frame=True)

print(cross_val_score(KNeighborsClassifier(), X, y, cv=5).mean())                                  # 0.928
# 스케일링 적용
print(cross_val_score(make_pipeline(StandardScaler(), KNeighborsClassifier()), X, y, cv=5).mean())  # 0.965
```
