```python
df = pd.read_csv("score.csv", encoding="utf-8")
```

```python
# 열 이름이 name인 데이터들을 조회
print(df["name"])
```

```python
# 열이름이 score, name인 데이터들을 조회 
print(df[["score", "name"]])
```

```python
df.iloc[1] # index가 1인 행을 선택
```

```python
df.iloc[0:2] #index가 0에서 2-1인 행들을 선택
```

```python
# name이 홍길동인 행을 조회
df[df["name"] == "홍길동"]
```

```python
# name이 홍길동이거나 성적이 80이상인 학생들을 조회
df[(df["name"] == "홍길동") | (df["score"] >= 80)]
```

```python
# pass 컬럼을 추가. 60이상이면 True 아니면 False
df["pass"] = df["score"] >= 60
df
```

```python
# 테스트용 컬럼을 추가
df["test"] = None
df["test2"] = None
print(df)
# 컬럼 삭제
df = df.drop("test", axis=1)
print(df)
df = df.drop(columns=["test2"])
print(df)
```

```python
# 결측치(값이 안들어 있음) 처리
# 결측치 확인
df.isna().sum()
```

```python
# name의 결측치를 "익명"으로 바꿈
df2 = df.fillna({"name" : "익명"})
df2
```
