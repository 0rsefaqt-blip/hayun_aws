```python
def 나이확인(age):
    if age >= 19:
        print(f"{age}는 성인입니다")
    else:
        print(f"{age}는 청소년입니다")

나이확인(19)


def 나이확인2(age):
    if age >= 19:
        return "성인"
    else:
        return "청소년"
age=20
print(f"{age}는 {나이확인2(age)}입니다")

def 나이확인3(age):
    if age >= 19:
        return True
    else:
        return False

나이확인3(age) # True

if 나이확인3(age):
    print(f"{age}살은 성인입니다.")
else:
    print(f"{age}살은 청소년입니다.")
```

```python
def 성적(score, cutline=60): # 함수에 기본값을 적용
    if cutline <= score <= 100:
        return "합격"
    elif 0 <= score < cutline:
        return "불합격"
    else:
        print("잘못된 입력입니다.")

#1
score = 80
cutline = 60
print(f"{score}점은 {성적(score)}입니다.") # 함수 호출 할 때 기본값이 있는 매개변수는 생략해도 됨
#2
성적(30)
```

```python
def to_dict(age, name="익명", address="주소없음"):
    return {"name" : name, "age" : age, "address" : address}

user1 = to_dict(21, "홍길동", "서울시")
user2 = to_dict(21, "홍길동")
user3 = to_dict(21)
user4 = to_dict(21, address="서울시")
print(user1)
print(user2)
print(user3)
print(user4)
```

```python
users = [
    {"name" : "홍길동", "age" : 21},
    {"name" : "임꺽정", "age" : 41},
    {"name" : "하니", "age" : 15},
]

def 함수(users):
    count = len(users)
    sum_age = 0
    for user in users:
        sum_age += user.get("age")
    mean_age = sum_age/count
    return count, mean_age

count, mean_age = 함수(users)
print(f"회원 {count}명의 평균 나이 {mean_age:.1f}")
```

```python
def 함수(users):
    count = len(users)
    ages = []
    for user in users:
        ages.appends(user.get("age"))
        
    sum_age = sum(ages)
    mean_age = sum_age/count
    return count, mean_age

count, mean_age = 함수(users)
print(f"회원 {count}명의 평균 나이 {mean_age:.1f}")
```
