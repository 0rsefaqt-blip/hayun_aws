## 조회, 필터, 정렬

### 행 조회
* iloc[n] : n번째 행을 선택
* iloc[m:n] : m번째 행부터 n-1번째 행을 선택
* set_index("컬럼명") : 인덱스를 해당 컬럼명으로 바꿈
* loc[n] : 인덱스가 n인 행을 선택
* loc["컬럼값A"] : 컬럼의 값이 A인 행을 선택

```python
print(df.iloc[0])
```

```python
print(df.iloc[0:2])
```

```python
df2 = df.set_index("name")
print(df2)
```

```python
print(df2.loc["홍길동"])
```

```python
print(df2.loc[["홍길동", "임꺽정"]])
```

### 필터
* 조건이 맞는 행을 조회
* df[ df을 이용한 조건 ]
* 조건이 여러개인 경우 &로 연결

```python
print(df[df["age"] >= 10])
```

```python
print(df[(df["age"] >= 10) & (df["age"] <= 30)])
```

### 정렬
* sort_values("컬럼명", ascending)
* ascending=True : 오름차순, ascending=False : 내림차순

```python
print(df.sort_values("name"))
```

```python
print(df.sort_values("name", ascending=False))
```

## 빈값(결측지) 처리

```python
df = pd.DataFrame({
    "name" : ["홍길동", "임꺽정", "하니", "고길동"],
    "score" : [100, 90, None, 80]
})
```

```python
print(df.isnull().sum()) # 컬럼별 빈 값 개수
```

```python
print(df.dropna())
```

```python
print(df.fillna(0))
```

```python
# 이름이 없으면 익명으로 바꾸고, 성적이 없으면 평균으로 수정
df2 = df["name"].fillna("익명")
df3 = df["score"].fillna(df["score"].mean())
df["name"] = df2
df["score"] = df3
print(df)
```

```python
df = pd.DataFrame({
    "name" : ["홍길동", "임꺽정", "하니", None],
    "score" : [100, 90, None, 80]
})
df4 = df.fillna({"name" : "익명", "score" : df["score"].mean()})
print(df4)
```

## 병합
* 컬럼이 서로 다른 프레임을 합치는 것

```python
std_df = pd.DataFrame({
    "id" : [1,2,4],
    "name" : ["홍길동", "임꺽정", "하니"]
})
score_df = pd.DataFrame({
    "id" : [1,2,3],
    "score" : [90, 70, 30]
})
m_df1 = pd.merge(std_df, score_df, on="id") # id를 기준으로 합침. 일치하는 id 없으면 버림
print(m_df1)
```

```python
m_df2 = pd.merge(std_df, score_df, on="id", how="left") # left join과 비슷
print(m_df2)
```

## 병합
* 위아래로 합침(쌓기)
* 두 데이터 프레임의 컬럼이 같고 데이터들을 합침

```python
# 1월 과일 판매량
df1 = pd.DataFrame({
    "month" : [1, 1],
    "item" : ["사과", "배"],
    "sale" : [10, 20]
})
# 2월 과일 판매량
df2 = pd.DataFrame({
    "month" : [2, 2],
    "item" : ["사과", "배"],
    "sale" : [30, 50]
})
```

```python
sale_df = pd.concat([df1, df2], ignore_index=True) # index를 무시
print(df1)
print(df2)
print(sale_df)
```

## 행 중복제거

```python
df = pd.DataFrame([
    {"name" : "홍길동", "age" : 20},
    {"name" : "홍길동", "age" : 40},
    {"name" : "임꺽정", "age" : 20},
    {"name" : "홍길동", "age" : 20},
])
print(df)
print(df.drop_duplicates()) # 크롤링해서 가져온 데이터가 중복 됐을때
```
