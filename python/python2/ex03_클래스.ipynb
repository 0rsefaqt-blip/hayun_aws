## 클래스 
* 필드 선언은 무조건 생성자에서만 가능 => X
* 접근제어자가 없음
* 클래스의 함수나 생성자는 반드시 첫번째 매개변수로 self를 넘겨주어야 함
```
class 클래스이름(부모클래스):

    변수 = 초기값 # 클래스 변수
    
    # 생성자 : 필드들을 선언 후에 초기화
    def __init__(self, 변수1, 변수2, ...):
        self.필드1 = 변수1 # 필드1은 자바의 인스턴스 변수(필드)
        
    # toString
    def __str__(self):
        return 필드를 포함한 문자열

    # 필요한 기능을 함수로 만듦
```

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"이름: {self.name}, 나이: {self.age}"

    # 기능 : 학생 정보를 콘솔에 출력하는 기능(print)
    # 이름 : 홍길동
    # 나이 : 17살
    def print(self):
        print(f"이름 : {self.name}")
        print(f"나이 : {self.age}살")
    
std1 = Student("홍길동", 17)
print(std1)
print(std1.__str__())
std1.print()
````

````python
from dataclasses import dataclass, field, asdict

@dataclass # 생성자를 자동으로 추가, __str__도 자동으로 추가
class Student2:
    name:str # 쿨래스 변수를 필드로 만듬
    age:int
    # 전화번호, 집전화, 폰번호
    numbs:list[str] =  field(default_factory=list) # []로 초기화 하지 않고 field(default_factory=list)로 빈 리스트 초기화

    def print(self):
        print(f"이름 : {self.name}")
        print(f"나이 : {self.age}살")

std2 = Student2("홍길동", 17)
std3 = Student2("임꺽정", 18)
print(std2)
print(std3)
print(asdict(std2)) # @dataclass를 이용하여 선언한 클래스의 객체를 dict로 변환
````

````python
# BankAccount 클래스를 만드세요.
# 계좌 정보 : owner(소유자), balance(잔액)
# 기능 : 입금 (deposit)
    # 입금액이 0 이하면 "금액은 0 보다 커야 합니다" 를 출력
# 기능 : 출금(withdraw)
    # 출금액이 0 이하면 "금액은 0 보다 커야 합니다" 를 출력
    # 잔액보다 출금액이 크면 "잔액이 부족합니다" 를 출력
# 기능 : 잔액확인(get_balance)
# 기능 : 거래내역 조회(print_history)
    # 입금 : 3000원
    # 출금 : 2000원

from dataclasses import dataclass, field, asdict

@dataclass
class BankAccount:
    owner: str
    balance: int
    history: list[str] = field(default_factory=list)
    
    def deposit(self, 돈):
        if 돈 <= 0:
            print("금액은 0보다 커야 합니다.")
        else:
            self.balance += 돈
            content = f"입금: {돈}원"
            self.history.append(content)
            print(content)

    def withdraw(self, 돈):
        if 돈 <= 0:
            print("금액은 0보다 커야 합니다.")
        if 돈 > self.balance:
            print("잔액이 부족합니다.")
        else:
            self.balance -= 돈
            content = f"출금: {돈}원"
            self.history.append(content)
            print(content)
    
    def get_balance(self):
        return self.balance

    def print_history(self):
        print("입출금내역")
        for item in self.history:
            print(item)

acc = BankAccount("홍길동", 10000)
acc.deposit(5000)
acc.deposit(-5000)
acc.withdraw(3000)
acc.withdraw(30000)

print(f"잔액 : {acc.get_balance()}")
acc.print_history()
    
````
