### 파일 읽고 쓰기
* 파일 열기
* `with open`은 파일을 열고 작업이 끝나면 파일을 자동으로 닫아줌
* 모드
    * `r`: 읽기 모드. 파일이 없으면 에러 발생
    * `w`: 쓰기 모드. 파일이 없으면 새 파일을 생성. 파일이 있으면 열어서 덮어씀
    * `a`: 추가 모드. 파일이 없으면 새 파일을 생성. 파일이 있으면 열어서 뒤에서 이어씀
    * `b`: 바이너리 파일
```
with open("파일명.확장자", "모드", encoding="utf-8") as f:
    # 파일 작업 코드
```

````python
with open("file.txt", "r", encoding="utf-8") as f:
    text = f.read()
    print(f"읽어온 내용: {text}")
````

````python
# 읽기 모드(r)에서는 예외처리를 하는 것이 좋다
try:
    with open("file1.txt", "r", encoding="utf-8") as f:
        text = f.read()
        print(f"읽어온 내용: {text}")
except FileNotFoundError as e:
    print(e)
````

````python
# 한 줄 씩 읽기
try:
    with open("file.txt", "r", encoding="utf-8") as f:
        for line in f:
            print(f"줄: {line}", end="") # end는 문자열 마지막에 넣을 문자
except FileNotFoundError as e:
    print(e)
````

````python
# 쓰기
with open("out.txt", "w", encoding="utf-8") as f:
    for i in range(1,6):
        f.write(str(i))
````

````python
# 내 은행 잔고를 파일에서 읽어와서 가져옴. 
# 파일명: bank_hong.txt
# 파일이 없으면 예외처리하고, 내 잔액은 0으로 설정
# 주의사항: 읽어온 값은 문자열. balance는 정수

try:
    with open("bank_hong.txt", "r", encoding="utf-8") as f:
        balance = int(f.read().strip()) # 읽어온 문자열을 정수로 변환
except FileNotFoundError as e:
    balance = 0
    
def deposit(balance, amount):
    if amount <= 0:
        raise ValueError("입금액은 0보다 커야 합니다.")
    return balance + amount

for amount in (1000, -1000, 10000):
    try:
        balance = deposit(balance, amount)
        print(f"입금 : {amount}원. 잔액: {balance}원")
        
    except ValueError as e:
        print(e)

# 내 은행 잔고에 변경된 잔고를 저장
# 파일명: bank_hong.txt
with open("bank_hong.txt", "w", encoding="utf-8") as f:
     f.write(str(balance))
````

### json 활용
* json 문자열 -> dict
    * json.loads() 
* dict -> json 문자열
    * json.dumps() 
* 파일에 dict을 저장하기
    * json.dump()
* 저장된 json 파일을 dict로 불러오기
    * json.load()

````python
import json
# dict
data = {
    "name" : "홍길동",
    "age" : 25,
    "car" : None
}
# dumps(dict, ensure_ascii)
# dict : 딕셔너리
# ensure_ascii : 아스키값들인지. 한글은 유니코드여서 False로 설정
data_str = json.dumps(data, ensure_ascii=False)
print(data, type(data))
print(data_str, type(data_str))

print("-----------------------------------------")
# json 형태의 문자열
str1 = '{"name" : "홍길동", "age" : 25, "car" : null}'
# Loads(str)
# str : json 문자열
str_data1 = json.loads(str1)
print("json => dict")
print(str1, type(str1))
print(str_data1, type(str_data1))
````

````python
try:
    with open("user.json", "r", encoding="utf-8") as f:
        users = json.load(f)
        print(users, type(users))
except FileNotFoundError as e:
    print(e)
# except Exception as e:
#     print(e)

data = {
    "name" : "홍길동",
    "age" : 25,
    "car" : None
}

with open("user.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
    print("저장이 완료됐습니다.")
````

````python
# 이름 : 홍길동, 나이: 25인 회원의 dict를 선언
user1 = {"name":"홍길동", "age":25}
# 이름: 임꺽정, 나이: 40인 회원의 dict를 선언
user2 = {"name":"임꺽정", "age":40}
# users 리스트에 홍길동 회원의 dict와 임꺽정 회원의 dict를 추가
users = []
users.append(user1)
users.append(user2)

# 회원 목록인 users를 users.json에 저장
with open("users.json", "w", encoding="utf-8") as f:
    json.dump(users, f, ensure_ascii=False, indent=2)
    print("저장이 완료됐습니다.")
````
